```python
import numpy as np
from scipy.fftpack import fft
import matplotlib.pyplot as plt

# Generate a square wave
fs = 20000  # Sampling frequency
T = 1/fs  # Sampling period
t = np.arange(0, 1, T)  # Time vector
f = 1000  # Frequency of the square wave
data = np.sign(np.sin(2*np.pi*f*t))  # Square wave generation

# Perform FFT
N = len(data)
yf = fft(data)
xf = np.fft.fftfreq(N, 1/fs)

# Find the peak frequency and amplitude
peak_index = np.argmax(np.abs(yf))
peak_freq = xf[peak_index]
peak_amplitude = np.abs(yf[peak_index])

print(f"Peak Frequency: {peak_freq} Hz")
print(f"Peak Amplitude: {peak_amplitude}")

# --- Changes start here ---
magnitudes = np.abs(yf)  # Get magnitudes of FFT output

# Find the indices of the top 3 peaks
peak_indices = np.argpartition(magnitudes, -5)[-5:]

# Sort the indices in descending order of magnitude
peak_indices = peak_indices[np.argsort(-magnitudes[peak_indices])]

# Get the frequencies and amplitudes of the top 3 peaks
peak_freqs = xf[peak_indices]
peak_amplitudes = magnitudes[peak_indices]

# Print the results
for i in range(5):
    print(f"{i+1}st Peak Frequency: {peak_freqs[i]} Hz")
    print(f"{i+1}st Peak Amplitude: {peak_amplitudes[i]}")
# --- Changes end here ---

# Plot the results
plt.plot(xf, np.abs(yf))
plt.title("FFT of Square Wave")
plt.xlabel("Frequency (Hz)")
plt.ylabel("Amplitude")
plt.xlim(0, fs/2)  # Show only positive frequencies
plt.show()
```
